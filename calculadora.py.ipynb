```python
# CALCULADORA

# num1, num2, operador, resultado

num1 = 0
num2 = 0
Op = ''
resultado = 0

print("CALCULADORA EBAC")
print("------------")


while True:
    try:

        num1 = float(input("Digite o primeiro número: "))
        op = input("Digite a operação matemática (+, -, *, /): ")
        num2 = float(input("Digite o segundo número: "))

        # Verifica se o operador é válido
        if op == "+":
            resultado = num1 + num2
        elif op == "/":
            resultado = num1 / num2
        elif op == "*":
            resultado = num1 * num2
        elif op == "-":
            resultado = num1 - num2
        else:
            print("Operador inválido! Digite uma operação válida (+, -, *, /).")
            continue # Retorna ao início do loop se o operador for inválido

        print(' {} {} {} = {}'.format(num1, op, num2, resultado))

    except ValueError:
        # Captura erros quando algo que não é um número é digitado
        print("Por favor, digite um número válido.")

    except ZeroDivisionError:
        # Captura o erro de divisão por zero
        print("Erro: Divisão por zero não permitida!")
```
